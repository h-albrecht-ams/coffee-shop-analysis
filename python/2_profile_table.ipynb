# Cluster Profiles: Prep for Tableau
Builds a normalized per-cluster profile table for the Tableau bar chart, exported as a separate CSV.

```python
import pandas as pd
```

```python
df = pd.read_csv("coffee_clusters_for_tableau.csv")
```

```python
df['cluster_name'].value_counts()
```

```python
#Average each feature per cluster: 1 row per cluster, showing its typical feature values

profiles = df.groupby('cluster_name')[['hour', 'transaction_qty', 'unit_price', 'is_weekend']].mean()
```

```python
profiles
```

```python
# Normalize features to 0–1: Features are on different scales (price ~21 vs weekend 0–1), raw averages would make price visually dominate the chart. 
# Min-max rescales each feature so clusters are comparable.

profiles_norm = (profiles - profiles.min()) / (profiles.max() - profiles.min())
```

```python
profiles_norm.round(2)
```

Values are *relative* (0 = lowest cluster, 1 = highest), not real units — a price of 0 means "lowest-priced cluster," not "free."

```python
# Drop hour and export
# Its real spread is under one hour, but normalization would stretch that to 0–1, making it visually dominate the chart.

profiles_norm = profiles_norm.drop(columns='hour')
profiles_norm.to_csv('cluster_profiles.csv')
```
