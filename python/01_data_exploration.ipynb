Calculate revenue (quantity transactions * unit price)

```python
df['revenue'] = df['transaction_qty'] * df['unit_price']

df['revenue'].sum()

round(df['revenue'].sum(), 2)
```

Calculate number of transactions per store location

```python
df['store_location'].value_counts()
```

Calculate revenue per store location

```python
df.groupby('store_location')['revenue'].sum()
```

Group by product category and sort by revenue

```python
df.groupby('product_category')['revenue'].sum().sort_values(ascending=False)
```
